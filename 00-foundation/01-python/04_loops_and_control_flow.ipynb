**Exercício 1: Laço for Básico**

Use um laço for para imprimir os números de 1 até 10.

```python
numeros = range(1, 11)

for numero in numeros:
    print(numero)
```

**Exercício 2: Laço while Básico**

Use um laço while para contar de 1 até 5 e exibir cada número.

```python
contador = 1

while contador <= 5:
  print(contador)
  contador = contador + 1
```

**Exercício 3: Calcular Média com for**

Use um laço for para calcular a média das notas: [7.5, 8.0, 9.5, 6.0, 8.5]

```python
notas = [7.5, 8.0, 9.5, 6.0, 8.5]
soma = 0

for nota in notas:
  soma = soma + nota

media = soma / len(notas)
print(f"Média: {media:.2f}")
```

**Exercício 4: Função range()**

Use range() e for para imprimir a tabuada do 7 (de 1 até 10).

```python
for numero in range(1, 11):
    print(f"7 x {numero} = {numero * 7}")
```

**Exercício 5: Números Pares com range()**

Use range() e for para imprimir todos os números pares de 0 até 20.

```python
for numero in range(0, 21, 2):
    print(numero)
```

**Exercício 6: Controle de Fluxo - break**

Crie uma lista de números: [5, 12, 8, 3, 15, 7]. Use um laço for para encontrar o primeiro número maior que 10 e parar quando encontrar.

```python
numeros = [5, 12, 8, 3, 15, 7]

for numero in numeros:
    if numero > 10:
        print(numero)
        break
```

**Exercício 7: Controle de Fluxo - continue**

Use continue para imprimir apenas os números positivos da lista: [-2, 5, -1, 8, -3, 10]

```python
numeros = [-2, 5, -1, 8, -3, 10]

for numero in numeros:
    if numero < 0:
        continue
    print(numero)
```

**Exercício 8: Iteração sobre Strings**

Crie um programa que itera sobre a string "Python" e conta quantas vogais existem.

```python
palavra = "Python"
vogais = "aeiouAEIOU"
contador_vogais = 0

for letra in palavra:
  if letra in vogais:
    contador_vogais = contador_vogais + 1

print(f"A quantidade de vogais na palavra {palavra} é {contador_vogais}.")
```

**Exercício 9: Sistema de Notas com Classificação**

Crie um programa que itera sobre uma lista de notas: [8.5, 6.0, 9.5, 7.0, 5.5] e classifica cada uma (>= 9.0: Excelente, >= 7.0: Bom, >= 5.0: Regular, < 5.0: Insuficiente).

```python
notas = [8.5, 6.0, 9.5, 7.0, 5.5]

for nota in notas:
    if nota >= 9:
        classificacao = "Excelente"
    elif nota >= 7:
        classificacao = "Bom"
    elif nota >= 5:
        classificacao = "Regular"
    else:
        classificacao = "Insuficiente"

    print(f"Nota: {nota:.2f} - {classificacao}")
```

**Exercício 10: Calculadora de Estatísticas**

Crie um programa que itera sobre uma lista de números: [15, 8, 23, 4, 42, 11, 7, 19] e calcula: soma, média, maior e menor valor, e conta quantos são pares e ímpares.

```python
lista = [15, 8, 23, 4, 42, 11, 7, 19]

soma = 0
pares = 0
impares = 0
maior = lista[0]
menor = lista[0]

for numero in lista:
  soma = soma + numero

  if numero > maior:
    maior = numero

  if numero < menor:
    menor = numero

  if numero % 2 == 0:
    pares = pares + 1
  else:
    impares = impares + 1

media = soma/len(lista)

print(f"Lista: {lista}")
print(f"Soma: {soma}")
print(f"Média: {media:.2f}")
print(f"Maior número: {maior}")
print(f"Menor número: {menor}")
print(f"Quantidade de pares: {pares}")
print(f"Quantidade de ímpares: {impares}")
```
