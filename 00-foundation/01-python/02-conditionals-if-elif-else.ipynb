## Objetivo do Projeto

## Parte 1: Cadastro e Validação de Usuário

Primeiro, vamos criar e validar os dados do usuário. O sistema verifica:

- Se a idade é adequada (mínimo 12 anos)
- Se o usuário possui carteira da biblioteca
- Se há multas pendentes

Dependendo do tipo de usuário (estudante ou regular), diferentes prazos de empréstimo são aplicados.

```python
idade = 15
carteira_biblioteca = True
multas = 2
tipo_usuario = "estudante"

if idade < 12:
    print("Você não possui idade para alugar livros nesta biblioteca.")

elif carteira_biblioteca == False:
    print("Você ainda não possui carteira da biblioteca.")

else:
    print("Usuário apto a utilizar a biblioteca.")

    if multas == 0:
      print("Você não possui multas.")
    else:
      print(f"Você possui {multas} multa(s) pendente(s).")

    if tipo_usuario == "estudante":
      prazo_emprestimo = 15
    else:
      prazo_emprestimo = 10

    print(f"Seu prazo de empréstimo é de {prazo_emprestimo} dias.")
```

## Parte 2: Sistema de Empréstimo

O sistema gerencia empréstimos de livros, considerando:

- O tipo de usuário (estudante tem prazo maior)
- Os dias desde o empréstimo
- Se o livro está em atraso ou ainda no prazo


```python
dias_emprestimo = 21
dias_faltantes = prazo_emprestimo - dias_emprestimo
dias_atraso = dias_emprestimo - prazo_emprestimo

if dias_emprestimo <= prazo_emprestimo:
    print(f"O prazo de entrega é em {dias_faltantes} dias.")
else:
    print(f"O livro está atrasado há {dias_atraso} dias.")
```

## Parte 3: Cálculo de Multa

O sistema calcula multas baseadas em:

- Dias de atraso
- Valor da multa por dia
- Acréscimo para atrasos grandes (> 7 dias)
- Desconto para estudantes (10%)


```python
valor_multa_dia = 2
valor_multa = dias_atraso * valor_multa_dia

if dias_atraso > 7:
  valor_multa = valor_multa * 1.5

if tipo_usuario == "estudante":
  valor_multa = valor_multa * 0.9

print(f"Valor da multa: R$ {valor_multa:.2f}")
```

## Parte 4: Sistema de Renovação

O sistema verifica se um livro pode ser renovado, considerando:

- Se o livro está em atraso
- Número de renovações já realizadas (máximo: 2)
- Se há reserva para o livro
- Se há multas pendentes

```python
livro_atrasado = False
numero_renovacoes = 1
reserva_livro = True
multas = 1

if livro_atrasado == True:
  print(f"Renovação não permitida: o livro está atrasado.")
elif numero_renovacoes >= 2:
  print(f"Renovação não permitida: limite de renovações atingido.")
elif reserva_livro == False:
  print(f"Renovação não permitida: não há reserva para o livro.")
elif multas > 0:
  print(f"Renovação não permitida: há {multas} multa(s) pendentes.")
else:
  print(f"Renovação permitida.")

```

### Tarefa:

Crie um sistema de gestão de loja que:

1. Valida dados do cliente (idade >= 18, tem cadastro)
2. Calcula desconto baseado em:
   - Valor da compra >= 100: 5%
   - Valor da compra >= 200: 10%
   - Valor da compra >= 500: 15%
   - Se for cliente VIP: +5% adicional
   - Desconto máximo: 25%
3. Calcula frete:
   - Distância < 50 km: R$ 10
   - Distância 50-200 km: R$ 20
   - Distância > 200 km: R$ 30
   - Se compra >= 200: frete grátis
4. Gera relatório resumido

### Dicas:

- Use apenas estruturas condicionais (`if`, `elif`, `else`)
- Use condições aninhadas quando necessário
- Nomes de variáveis descritivos

```python
idade = 19
valor_compra = 538.75
distancia = 10
cliente_vip = True

if idade < 18:
    print("Você precisa ter ao menos 18 anos para realizar compras nessa loja.")
else:
    print("Dados validados.")

if valor_compra >= 500:
    desconto = 0.15
elif valor_compra >= 200:
    desconto = 0.10
elif valor_compra >= 100:
    desconto = 0.05
else:
    desconto = 0

if cliente_vip:
    desconto = desconto + 0.05

if desconto > 0.25:
    desconto = 0.25

valor_desconto = valor_compra * desconto
valor_com_desconto = valor_compra - valor_desconto

if valor_compra >= 200:
    frete = 0
elif distancia > 200:
    frete = 30
elif distancia >= 50:
    frete = 20
else:
    frete = 10

total_compra = valor_com_desconto + frete

print(f"Valor original: R$ {valor_compra:.2f}")
print(f"Desconto: R$ {valor_desconto:.2f}")
print(f"Frete: R$ {frete:.2f}")
print(f"Total da compra: R$ {total_compra:.2f}")
```
