**Exercício 1: Estrutura if Básica**

Crie um programa que verifica se uma pessoa é maior de idade. Se a idade for >= 18, exiba "Você é maior de idade". Use idade = 20.

Crie um programa que verifica se um número é par ou ímpar. Se o número for par (resto da divisão por 2 é 0), exiba "Par", caso contrário, exiba "Ímpar". Use numero = 17.

**Exercício 5: Condições Aninhadas - Sistema de Aprovação**

Crie um programa que verifica aprovação:

1. Se nota >= 7.0, está aprovado
2. Se aprovado, verifique se nota >= 9.0 (excelente) ou >= 8.0 (muito bom)
3. Se reprovado, verifique se nota >= 5.0 (pode recuperar) ou < 5.0 (reprovado)

Use nota = 8.5.

```python
nota = 8.5

if nota >= 7.0:
  print(f"Nota: {nota} - Aprovado")
  if nota >= 9.0:
    print("Excelente!")
  elif nota >= 8.0:
    print("Muito bom!")
else:
  print(f"Nota: {nota} - Reprovado")
  if nota >= 5.0:
    print("Recuperação.")
  else:
    print("Refazer a disciplina.")
```

**Exercício 6: Sistema de Desconto**

Crie um programa que calcula desconto baseado no valor da compra:

* = 200: 20% de desconto

* = 100: 10% de desconto

* = 50: 5% de desconto

* < 50: sem desconto

Calcule e exiba o valor final. Use valor_compra = 150.00.

```python
valor_compra = 150.00

if valor_compra >= 200:
    desconto = 0.20
elif valor_compra >= 100:
    desconto = 0.10
elif valor_compra >= 50:
    desconto = 0.05
else:
    desconto = 0

desconto_compra = valor_compra * desconto
valor_final = valor_compra - desconto_compra

print(f"Valor original: R$ {valor_compra:.2f}")
print(f"Desconto: {desconto:.0%}")
print(f"Valor final: R$ {valor_final:.2f}")
```

**Exercício 7: Sistema de Votação**

Crie um programa que verifica se uma pessoa pode votar:

* < 16 anos: não pode votar
* 16-17 anos: voto opcional
* 18-70 anos: voto obrigatório
* 70 anos: voto opcional

Use idade = 25.

**Exercício 8: Sistema de Frete com Condições Aninhadas**

Crie um programa que calcula frete:

* Distância < 50 km: R\$ 10,00
* Distância 50-200 km: R\$ 15,00
* Distância > 200 km: R\$ 20,00

Além disso:

* Se peso > 10 kg: adicione R\$ 5,00
* Se for cliente VIP: desconte 10% do frete total

Use distancia = 150 km, peso = 12 kg e e_cliente_vip = True.

```python
distancia = 150
peso = 12
e_cliente_vip = True

if distancia > 200:
    frete = 20
elif distancia >= 50:
    frete = 15
else:
    frete = 10

if peso > 10:
    frete = frete + 5

if e_cliente_vip:
    frete = frete - (frete * 0.10)

print(f"Frete: R${frete:.2f}")
```

**Exercício 9: Projeto Integrado - Sistema de Empréstimo**

Crie um sistema de avaliação de empréstimo que:

1. Verifica se tem restrições no CPF (se sim, negado independente do salário)
2. Se salário >= R\$ 3.000: aprovado
3. Se salário entre R\$ 2.000 e R$ 2.999: análise especial
4. Se salário < R\$ 2.000: negado
5. Se for cliente há mais de 2 anos: aumenta chance de aprovação

Use salario = 3500.00, tem_restricoes = False e tempo_cliente = 3 anos.

```python
salario = 3500.00
tem_restricoes = False
tempo_cliente = 3

if tem_restricoes:
    print("Status: Empréstimo negado por restrições no CPF.")

else:
    if salario >= 3000:
        print("Status: Empréstimo aprovado.")

    elif salario >= 2000:
        print("Status: Empréstimo em análise.")

    else:
        print("Status: Empréstimo negado.")
        print("Salário mínimo: R$ 2.000,00")

    if tempo_cliente > 2:
        print("Cliente antigo - Maior chance de aprovação.")
```
