**Exercício 3:** Operadores Aritméticos  

Um produto custa R$ 89.90. Você tem um desconto de 15% e precisa calcular:

**1.**   O valor do desconto  
**2.**   O preço final após o desconto  
**3.**   Se você comprar 3 unidades, quanto pagará no total?  

```python
preco_original = 89.90
percentual_desconto = 15
valor_desconto = preco_original * (percentual_desconto / 100)
preco_final = preco_original - valor_desconto
valor_03_unidades = preco_final * 3
print(f"Valor do Desconto de {percentual_desconto}%: {valor_desconto:.2f}")
print(f"Preço Final com Desconto: {preco_final:.2f}")
print(f"Valor de 3 Unidades: {valor_03_unidades:.2f}")
```

**Exercício 9:** F-strings com Formatação de Números

Um produto custa R\$ 99.98765. Formate o preço para:

- Mostrar com 2 casas decimais: R\$ 99.99
- Mostrar sem casas decimais: R\$ 100
- Calcular 10% de desconto e mostrar o valor final formatado.

```python
valor = 99.98765
duas_decimais = f"{valor:.2f}"
sem_decimais = f"{valor:.0f}"
desconto_percentual = valor * 0.10
desconto_valor = valor - desconto_percentual
print(duas_decimais)
print(sem_decimais)
print(f"{desconto_valor:.2f}")
```

*   Limpe e formate o nome do produto (remova espaços e use title())

*   Calcule o valor total (preço × quantidade)
